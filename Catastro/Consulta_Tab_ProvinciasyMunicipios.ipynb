###Creamos un archivo .pkl para almacenar la informacion y otro CSV

Provincias

```python
import pandas as pd
import pickle
from pycatastro import PyCatastro

if __name__ == "__main__":
    resultado_Provincias = PyCatastro.ConsultaProvincia()

provincias = resultado_Provincias['consulta_provinciero']['provinciero']['prov']

tabla_data = [{'Código Provincia': provincia['cpine'], 'Nombre Provincia': provincia['np']} for provincia in provincias]

Tabla_Provincias = pd.DataFrame(tabla_data)

# Ruta donde guardar el archivo .pkl
ruta_guardado = '/content/drive/MyDrive/Catastro/BD_Catastro_Actual/Tabla_Provincias.pkl'

# Ruta donde guardar el archivo .csv
ruta_csv = '/content/drive/MyDrive/Catastro/BD_Catastro_Actual/Tabla_Provincias.csv'

# Guardar la tabla en un archivo .pkl
with open(ruta_guardado, 'wb') as archivo:
    pickle.dump(Tabla_Provincias, archivo)

# Guardar la tabla en formato .csv
Tabla_Provincias.to_csv(ruta_csv, index=False)

# Imprimir la tabla
#print(tabla_Provincias)
```

Municipios

```python
import pandas as pd
import pickle
from pycatastro import PyCatastro

if __name__ == "__main__":
    # Código de provincia para la que queremos obtener los municipios
    codigo_provincia = "ASTURIAS"

    # Realiza la consulta de municipios para la provincia específica
    resultado_Municipios = PyCatastro.ConsultaMunicipio(codigo_provincia)

    # Extraer la lista de municipios del diccionario resultado
    municipios = resultado_Municipios['consulta_municipiero']['municipiero']['muni']

    # Crear una lista de diccionarios para la tabla
    tabla_data = [{'Código Municipio': municipio['loine']['cm'], 'Nombre Municipio': municipio['nm']} for municipio in municipios]

    # Crear la tabla utilizando pandas
    Tabla_Municipios = pd.DataFrame(tabla_data)

    # Ruta donde guardar el archivo .pkl
    ruta_pkl = '/content/drive/MyDrive/Catastro/BD_Catastro_Actual/Tabla_Municipios.pkl'

    # Ruta donde guardar el archivo .csv
    ruta_csv = '/content/drive/MyDrive/Catastro/BD_Catastro_Actual/Tabla_Municipios.csv'

    # Guardar la tabla en formato .pkl
    with open(ruta_pkl, 'wb') as archivo_pkl:
        pickle.dump(Tabla_Municipios, archivo_pkl)

    # Guardar la tabla en formato .csv
    Tabla_Municipios.to_csv(ruta_csv, index=False)

    # Imprimir la tabla
    #print(Tabla_Municipios)
```
