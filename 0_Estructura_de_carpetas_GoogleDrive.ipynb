## 2. Copia y Renombrado de Archivos de Plantilla

Esta sección contiene las funciones y la lógica para copiar archivos de plantilla desde una ubicación central (`Semilla_Raw`) a las carpetas específicas del monte que acabas de crear. Se han implementado checks para evitar sobrescribir archivos existentes y para asegurar que los archivos de origen realmente existan.

**¡Importante!** Asegúrate de que los archivos de plantilla estén presentes en `/content/drive/MyDrive/Catastro/General/Semilla_Raw/` antes de ejecutar esta celda.

```python
# --- Funciones auxiliares para copiar y renombrar ---
def copy_file_if_not_exists(source_path, dest_dir, dest_filename):
    """Copia un archivo si el archivo de origen existe y el archivo de destino no."""
    if not os.path.exists(source_path):
        print(f"ERROR: El archivo de origen no se encontró en: {source_path}")
        print("Por favor, verifica la ruta y asegúrate de que el archivo existe en tu Google Drive.")
        return

    dest_path = os.path.join(dest_dir, dest_filename)
    if os.path.exists(dest_path):
        print(f"Advertencia: El archivo '{dest_filename}' ya existe en '{dest_dir}'. No se copiará para evitar sobrescribir.")
    else:
        try:
            shutil.copyfile(source_path, dest_path)
            print(f"Archivo '{dest_filename}' copiado a '{dest_dir}'.")
        except Exception as e:
            print(f"ERROR al copiar '{dest_filename}' a '{dest_dir}': {e}")

def rename_file_if_exists(old_path, new_path):
    """Renombra un archivo si el archivo de origen existe y el nuevo nombre no existe."""
    if not os.path.exists(old_path):
        print(f"ERROR: No se puede renombrar. El archivo '{os.path.basename(old_path)}' no se encontró en: {old_path}")
    elif os.path.exists(new_path):
        print(f"Advertencia: El archivo '{os.path.basename(new_path)}' ya existe en el destino. No se renombrará para evitar sobrescribir.")
    else:
        try:
            os.rename(old_path, new_path)
            print(f"Archivo '{os.path.basename(old_path)}' renombrado a '{os.path.basename(new_path)}'.")
        except Exception as e:
            print(f"ERROR al renombrar '{os.path.basename(old_path)}' a '{os.path.basename(new_path)}': {e}")

# --- Copiar y Renombrar Archivos ---
# Archivo 1: Semilla.xlsx
filestocopy_1 = "/content/drive/MyDrive/Catastro/General/Semilla_Raw/Semilla.xlsx"
dir_destino_1 = dirConsulta_monte_1_1 + "/"
copy_file_if_not_exists(filestocopy_1, dir_destino_1, "Semilla.xlsx")
sinrenombrar_1 = dir_destino_1 + "Semilla.xlsx"
renombrado_1 = dir_destino_1 + "Semilla_" + monte + ".xlsx"
rename_file_if_exists(sinrenombrar_1, renombrado_1)

# Archivo 2: Plantilla_ConsultaMasiva.txt
filestocopy_2 = "/content/drive/MyDrive/Catastro/General/Semilla_Raw/Plantilla_ConsultaMasiva.txt"
dir_destino_2 = dirConsulta_monte_1_1 + "/"
copy_file_if_not_exists(filestocopy_2, dir_destino_2, "ConsultaMasiva.txt")
sinrenombrar_2 = dir_destino_2 + "ConsultaMasiva.txt"
renombrado_2 = dir_destino_2 + "ConsultaMasiva_" + monte + ".txt"
rename_file_if_exists(sinrenombrar_2, renombrado_2)

# Archivo 3: Plantilla_en_Blanco.xlsx
filestocopy_3 = "/content/drive/MyDrive/Catastro/General/Semilla_Raw/Plantilla_en_Blanco.xlsx"
dir_destino_3 = dirConsulta_monte_2 + "/"
copy_file_if_not_exists(filestocopy_3, dir_destino_3, "RC_" + monte + ".xlsx")
sinrenombrar_3 = dir_destino_3 + "RC_" + monte + ".xlsx"
renombrado_3 = dir_destino_3 + "Plantilla_en_Blanco_" + monte + ".xlsx"
rename_file_if_exists(sinrenombrar_3, renombrado_3)
```
